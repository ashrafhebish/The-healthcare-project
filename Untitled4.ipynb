```python
# PHASE 1 — DATA COLLECTION & PREPARATION

import pandas as pd
import numpy as np

file_path = "healthcare-dataset-stroke-data (4).csv"
df = pd.read_csv(file_path)

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

numerical_columns = [
    "age",
    "avg_glucose_level",
    "bmi"
]

categorical_columns = [
    "gender",
    "hypertension",
    "heart_disease",
    "ever_married",
    "work_type",
    "Residence_type",
    "smoking_status",
    "stroke"
]

print("\nNumerical Variables:")
print(numerical_columns)

print("\nCategorical Variables:")
print(categorical_columns)

print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

invalid_age = (df["age"] <= 0).sum()
invalid_glucose = (df["avg_glucose_level"] <= 0).sum()
invalid_bmi = (df["bmi"] <= 0).sum()

print("\nInvalid Data Checks:")
print(f"Invalid age values: {invalid_age}")
print(f"Invalid glucose values: {invalid_glucose}")
print(f"Invalid BMI values: {invalid_bmi}")

df_clean = df.copy()

df_clean["bmi"] = df_clean["bmi"].replace("N/A", np.nan).astype(float)

df_clean = df_clean.drop_duplicates()

df_clean.loc[df_clean["age"] <= 0, "age"] = np.nan
df_clean.loc[df_clean["avg_glucose_level"] <= 0, "avg_glucose_level"] = np.nan
df_clean.loc[df_clean["bmi"] <= 0, "bmi"] = np.nan

missing_bmi_count = df_clean["bmi"].isnull().sum()
bmi_median = df_clean["bmi"].median()

print("\nBMI Cleaning:")
print("Missing BMI values before cleaning:", missing_bmi_count)
print("BMI median used for imputation:", bmi_median)

df_clean["bmi"] = df_clean["bmi"].fillna(bmi_median)

for column in ["age", "avg_glucose_level"]:
    if df_clean[column].isnull().sum() > 0:
        df_clean[column] = df_clean[column].fillna(
            df_clean[column].median()
        )

print("\nMissing Values After Cleaning:")
print(df_clean.isnull().sum())

print("\nDuplicate Rows After Cleaning:")
print(df_clean.duplicated().sum())

print("\nFinal Dataset Shape:", df_clean.shape)

print("\nInvalid Values After Cleaning:")
print("Invalid age values:", (df_clean["age"] <= 0).sum())
print("Invalid glucose values:", (df_clean["avg_glucose_level"] <= 0).sum())
print("Invalid BMI values:", (df_clean["bmi"] <= 0).sum())
```

```python
# PHASE 2 — DESCRIPTIVE STATISTICS

numerical_columns = ["age", "avg_glucose_level", "bmi"]

descriptive_stats = pd.DataFrame(index=numerical_columns)

descriptive_stats["Mean"] = df_clean[numerical_columns].mean()

descriptive_stats["Median"] = df_clean[numerical_columns].median()


mode_values = df_clean[numerical_columns].mode()
for col in numerical_columns:
    if not mode_values[col].empty:
        descriptive_stats.loc[col, "Mode"] = mode_values[col].iloc[0]
    else:
        descriptive_stats.loc[col, "Mode"] = np.nan


descriptive_stats["Range"] = (
    df_clean[numerical_columns].max()
    - df_clean[numerical_columns].min()
)


descriptive_stats["Variance"] = df_clean[numerical_columns].var()


descriptive_stats["Standard Deviation"] = (
    df_clean[numerical_columns].std()
)


descriptive_stats["Q1"] = df_clean[numerical_columns].quantile(0.25)
descriptive_stats["Q2"] = df_clean[numerical_columns].quantile(0.50)
descriptive_stats["Q3"] = df_clean[numerical_columns].quantile(0.75)


descriptive_stats["IQR"] = (
    descriptive_stats["Q3"]
    - descriptive_stats["Q1"]
)


print("DESCRIPTIVE STATISTICS")
display(descriptive_stats.round(2))


print("\nINTERPRETATION")

for column in numerical_columns:

    mean = descriptive_stats.loc[column, "Mean"]
    median = descriptive_stats.loc[column, "Median"]
    mode = descriptive_stats.loc[column, "Mode"]
    std = descriptive_stats.loc[column, "Standard Deviation"]
    variance = descriptive_stats.loc[column, "Variance"]
    data_range = descriptive_stats.loc[column, "Range"]
    q1 = descriptive_stats.loc[column, "Q1"]
    q2 = descriptive_stats.loc[column, "Q2"]
    q3 = descriptive_stats.loc[column, "Q3"]
    iqr = descriptive_stats.loc[column, "IQR"]

    print(f"\n{column}:")
    print(f"- Mean: {mean:.2f}")
    print(f"- Median: {median:.2f}")
    print(f"- Mode: {mode:.2f}")
    print(f"- Range: {data_range:.2f}")
    print(f"- Variance: {variance:.2f}")
    print(f"- Standard Deviation: {std:.2f}")
    print(f"- Q1: {q1:.2f}")
    print(f"- Q2 (Median): {q2:.2f}")
    print(f"- Q3: {q3:.2f}")
    print(f"- IQR: {iqr:.2f}")

    
    if mean > median:
        print("- Interpretation: The mean is higher than the median, suggesting a right-skewed distribution.")
    elif mean < median:
        print("- Interpretation: The mean is lower than the median, suggesting a left-skewed distribution.")
    else:
        print("- Interpretation: The mean and median are approximately equal, suggesting a relatively symmetric distribution.")

    print(
        f"- The middle 50% of the data lies between "
        f"{q1:.2f} and {q3:.2f}."
    )

    print(
        f"- The standard deviation is {std:.2f}, "
        f"showing the typical spread of values around the mean."
    )
```

```python
# PHASE 3 — DISTRIBUTION ANALYSIS
import matplotlib.pyplot as plt

numerical_columns = ["age", "avg_glucose_level", "bmi"]

# 1. Frequency Distribution
for column in numerical_columns:

    print(f"\n===== Frequency Distribution: {column} =====")

    frequency_table = pd.cut(
        df_clean[column],
        bins=10
    ).value_counts().sort_index()

    display(frequency_table)

# 2. Histograms
for column in numerical_columns:

    plt.figure(figsize=(8, 5))

    plt.hist(
        df_clean[column],
        bins=10,
        edgecolor="black"
    )

    plt.title(f"Histogram of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.grid(axis="y", alpha=0.3)

    plt.show()


# 3. Box Plots
for column in numerical_columns:

    plt.figure(figsize=(8, 4))

    plt.boxplot(
        df_clean[column],
        vert=False
    )

    plt.title(f"Box Plot of {column}")
    plt.xlabel(column)

    plt.show()

# 4. Outlier Detection using IQR
outlier_summary = []

for column in numerical_columns:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[column] < lower_bound) |
        (df_clean[column] > upper_bound)
    ]

    outlier_summary.append({
        "Variable": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Number of Outliers": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_summary)

print("\nOUTLIER SUMMARY")
display(outlier_summary.round(2))

# 5. Distribution Shape using Skewness
print("\nDISTRIBUTION SHAPE")

for column in numerical_columns:

    skewness = df_clean[column].skew()
    mean = df_clean[column].mean()
    median = df_clean[column].median()

    print(f"\n{column}:")
    print(f"- Skewness = {skewness:.3f}")
    print(f"- Mean = {mean:.2f}")
    print(f"- Median = {median:.2f}")

    if skewness > 0.5:
        print("- Distribution Shape: Positively (right) skewed.")
    elif skewness < -0.5:
        print("- Distribution Shape: Negatively (left) skewed.")
    else:
        print("- Distribution Shape: Approximately symmetric.")

    if mean > median:
        print("- Mean is greater than the median.")
    elif mean < median:
        print("- Mean is less than the median.")
    else:
        print("- Mean and median are approximately equal.")
```

```python
# PHASE 4 — LINEAR ALGEBRA


from sklearn.preprocessing import MinMaxScaler

numerical_columns = ["age", "avg_glucose_level", "bmi"]

# 1. VECTORS
stroke_profile = df_clean[
    df_clean["stroke"] == 1
][numerical_columns].mean()

healthy_profile = df_clean[
    df_clean["stroke"] == 0
][numerical_columns].mean()

print("STROKE PROFILE VECTOR")
print(stroke_profile.round(2))

print("\nNO-STROKE PROFILE VECTOR")
print(healthy_profile.round(2))


# 2. SCALAR MULTIPLICATION
scenario_10_percent_worse = stroke_profile * 1.10
scenario_10_percent_better = stroke_profile * 0.90

print("\nSCALAR MULTIPLICATION")

print("\nStroke profile × 1.10 (10% increase):")
print(scenario_10_percent_worse.round(2))

print("\nStroke profile × 0.90 (10% decrease):")
print(scenario_10_percent_better.round(2))


# 3. DOT PRODUCT
scaler = MinMaxScaler()

normalized_data = scaler.fit_transform(
    df_clean[numerical_columns]
)

normalized_df = pd.DataFrame(
    normalized_data,
    columns=numerical_columns
)

# Get normalized profiles
normalized_stroke_profile = normalized_df[
    df_clean["stroke"] == 1
].mean()

normalized_healthy_profile = normalized_df[
    df_clean["stroke"] == 0
].mean()

# Risk weights
risk_weights = np.array([0.50, 0.30, 0.20])

stroke_dot_product = np.dot(
    normalized_stroke_profile.values,
    risk_weights
)

healthy_dot_product = np.dot(
    normalized_healthy_profile.values,
    risk_weights
)

print("\nDOT PRODUCT")

print("Risk weights:")
print(risk_weights)

print(
    f"\nStroke profile risk score: "
    f"{stroke_dot_product:.2f}"
)

print(
    f"No-stroke profile risk score: "
    f"{healthy_dot_product:.2f}"
)


# 4. MATRIX
# Rows = smoking status
# Columns = mean age, glucose and BMI

group_matrix = df_clean.groupby(
    "smoking_status"
)[numerical_columns].mean()

print("\nGROUP MATRIX — SMOKING STATUS")
display(group_matrix.round(2))

print(
    f"\nMatrix Shape: "
    f"{group_matrix.shape[0]} × {group_matrix.shape[1]}"
)
```

```python
# PHASE 5 — PROBABILITY

# 1. BASIC PROBABILITY
total_patients = len(df_clean)

# P(Stroke)
p_stroke = (
    (df_clean["stroke"] == 1).sum()
    / total_patients
)

print("BASIC PROBABILITY")
print(f"P(Stroke) = {p_stroke:.2%}")


# 2. CONDITIONAL PROBABILITY
# P(Stroke | Hypertension)
hypertension_patients = df_clean[
    df_clean["hypertension"] == 1
]

p_stroke_given_hypertension = (
    (hypertension_patients["stroke"] == 1).sum()
    / len(hypertension_patients)
)

# P(Stroke | Heart Disease)
heart_disease_patients = df_clean[
    df_clean["heart_disease"] == 1
]

p_stroke_given_heart_disease = (
    (heart_disease_patients["stroke"] == 1).sum()
    / len(heart_disease_patients)
)

# P(Stroke | Age >= 60)
senior_patients = df_clean[
    df_clean["age"] >= 60
]

p_stroke_given_senior = (
    (senior_patients["stroke"] == 1).sum()
    / len(senior_patients)
)

print("\nCONDITIONAL PROBABILITY")

print(
    f"P(Stroke | Hypertension) = "
    f"{p_stroke_given_hypertension:.2%}"
)

print(
    f"P(Stroke | Heart Disease) = "
    f"{p_stroke_given_heart_disease:.2%}"
)

print(
    f"P(Stroke | Age >= 60) = "
    f"{p_stroke_given_senior:.2%}"
)


# 3. EXPECTED VALUE
p_no_stroke = (
    (df_clean["stroke"] == 0).sum()
    / total_patients
)

expected_stroke = (
    0 * p_no_stroke +
    1 * p_stroke
)

print("\nEXPECTED VALUE")

print(
    f"E(Stroke) = "
    f"{expected_stroke:.4f}"
)
```

```python
# PHASE 6 — SAMPLING & STATISTICAL INFERENCE

# Reproducible random sampling
np.random.seed(42)

# Variable selected for sampling
sampling_column = "avg_glucose_level"

# 1. DEFINE THE POPULATION
population = df_clean[sampling_column]

population_size = len(population)

population_mean = population.mean()
population_std = population.std(ddof=0)

print("POPULATION")
print(f"Population size (N): {population_size}")
print(f"Population mean: {population_mean:.2f}")
print(f"Population standard deviation: {population_std:.2f}")

# 2. SELECT A REPRESENTATIVE SAMPLE
sample_size = 100

sample_parts = []

for stroke_value, group in df_clean.groupby("stroke"):

    proportion = len(group) / len(df_clean)

    group_sample_size = max(1, round(
        sample_size * proportion
    ))

    group_sample = group.sample(
        n=group_sample_size,
        random_state=42
    )

    sample_parts.append(group_sample)

sample_df = pd.concat(sample_parts)

if len(sample_df) > sample_size:

    sample_df = sample_df.sample(
        n=sample_size,
        random_state=42
    )

elif len(sample_df) < sample_size:

    remaining = df_clean.drop(
        sample_df.index
    )

    extra = remaining.sample(
        n=sample_size - len(sample_df),
        random_state=42
    )

    sample_df = pd.concat(
        [sample_df, extra]
    )

print("\nSAMPLE")
print(f"Sample size (n): {len(sample_df)}")

print("\nStroke distribution in Population:")
print(
    df_clean["stroke"]
    .value_counts(normalize=True)
    .round(3)
)

print("\nStroke distribution in Sample:")
print(
    sample_df["stroke"]
    .value_counts(normalize=True)
    .round(3)
)

# 3. SAMPLE MEAN
sample_mean = sample_df[sampling_column].mean()

print("\nSAMPLE MEAN")
print(
    f"Sample mean of {sampling_column}: "
    f"{sample_mean:.2f}"
)

# 4. SAMPLE STANDARD DEVIATION
sample_std = sample_df[sampling_column].std(
    ddof=1
)

print("\nSAMPLE STANDARD DEVIATION")
print(
    f"Sample standard deviation: "
    f"{sample_std:.2f}"
)

# 5. STANDARD ERROR
standard_error = (
    sample_std / np.sqrt(len(sample_df))
)

print("\nSTANDARD ERROR")

print(
    f"SE = s / √n = "
    f"{standard_error:.2f}"
)

# 6. COMPARE SAMPLE AND POPULATION
sampling_error = (
    sample_mean - population_mean
)

absolute_error = abs(
    sampling_error
)

percent_error = (
    absolute_error
    / population_mean
    * 100
)

print("\nSAMPLE VS POPULATION")

print(
    f"Population mean = "
    f"{population_mean:.2f}"
)

print(
    f"Sample mean = "
    f"{sample_mean:.2f}"
)

print(
    f"Sampling error = "
    f"{sampling_error:.2f}"
)

print(
    f"Percentage error = "
    f"{percent_error:.2f}%"
)

# 7. STATISTICAL INFERENCE
print("\nSTATISTICAL INFERENCE")

if absolute_error <= 1.96 * standard_error:

    print(
        "The sample mean is reasonably close to "
        "the population mean."
    )

    print(
        "The sample provides a reasonable "
        "representation of the population."
    )

else:

    print(
        "The sample mean differs noticeably "
        "from the population mean."
    )

    print(
        "A larger or different sample may "
        "provide a better estimate."
    )

# 8. SUMMARY TABLE
sampling_summary = pd.DataFrame({
    "Population Size": [population_size],
    "Sample Size": [len(sample_df)],
    "Population Mean": [population_mean],
    "Sample Mean": [sample_mean],
    "Sample Std": [sample_std],
    "Standard Error": [standard_error],
    "Sampling Error": [sampling_error],
    "Percentage Error": [percent_error]
})

print("\nSAMPLING SUMMARY")

display(
    sampling_summary.round(2)
)
```

```python
# PHASE 7 — CONFIDENCE INTERVALS

# Variable selected for confidence interval
from scipy import stats
confidence_column = "avg_glucose_level"


sample_values = sample_df[confidence_column]


confidence_level = 0.95
alpha = 1 - confidence_level


sample_mean = sample_values.mean()
sample_std = sample_values.std(ddof=1)
n = len(sample_values)


standard_error = sample_std / np.sqrt(n)


t_critical = stats.t.ppf(
    1 - alpha / 2,
    df=n - 1
)


margin_of_error = (
    t_critical * standard_error
)

# Confidence Interval
lower_bound = sample_mean - margin_of_error
upper_bound = sample_mean + margin_of_error


print("95% CONFIDENCE INTERVAL")
print("-------------------------------------")

print(f"Variable: {confidence_column}")
print(f"Confidence Level: {confidence_level:.0%}")
print(f"Sample Size: {n}")
print(f"Sample Mean: {sample_mean:.2f}")
print(f"Sample Standard Deviation: {sample_std:.2f}")
print(f"Standard Error: {standard_error:.2f}")
print(f"t-critical: {t_critical:.4f}")
print(f"Margin of Error: {margin_of_error:.2f}")

print(
    f"\n95% Confidence Interval: "
    f"({lower_bound:.2f}, {upper_bound:.2f})"
)


print("\nINTERPRETATION")

print(
    f"We are 95% confident that the true population mean "
    f"of average glucose level is between "
    f"{lower_bound:.2f} and {upper_bound:.2f}."
)
```

```python
# PHASE 8 — HYPOTHESIS TESTING

test_column = "avg_glucose_level"

# Create two groups
stroke_group = df_clean[
    df_clean["stroke"] == 1
][test_column]

no_stroke_group = df_clean[
    df_clean["stroke"] == 0
][test_column]


print("HYPOTHESIS TESTING")
print("-------------------------------------")

print("H0: The mean glucose level is the same")
print("    for stroke and no-stroke patients.")

print("\nH1: The mean glucose level is different")
print("    between stroke and no-stroke patients.")

# Significance level
alpha = 0.05

print(f"\nSignificance Level (α): {alpha}")


t_statistic, p_value = stats.ttest_ind(
    stroke_group,
    no_stroke_group,
    equal_var=False
)

print("\nTEST RESULTS")
print("-------------------------------------")

print(f"Stroke group mean: {stroke_group.mean():.2f}")
print(f"No-stroke group mean: {no_stroke_group.mean():.2f}")
print(f"t-statistic: {t_statistic:.4f}")
print(f"p-value: {p_value:.6f}")


print("\nDECISION")

if p_value < alpha:
    print("Reject H0.")
else:
    print("Fail to reject H0.")


print("\nCONCLUSION")

if p_value < alpha:
    print(
        "There is statistically significant evidence that "
        "the mean glucose level differs between stroke "
        "and no-stroke patients."
    )
else:
    print(
        "There is not enough statistical evidence to conclude "
        "that the mean glucose level differs between stroke "
        "and no-stroke patients."
    )
```

```python
# PHASE 9 — ONE-WAY ANOVA

# Variable selected for ANOVA
anova_column = "avg_glucose_level"


smoking_groups = []

group_names = []

for name, group in df_clean.groupby("smoking_status"):

    values = group[anova_column].dropna()

    smoking_groups.append(values)
    group_names.append(name)

# 1. GROUP MEANS
print("ONE-WAY ANOVA")
print("-------------------------------------")

print("Mean Average Glucose Level by Smoking Status:\n")

for name, values in zip(group_names, smoking_groups):

    print(
        f"{name}: "
        f"{values.mean():.2f}"
    )

# 2. HYPOTHESES
print("\nHYPOTHESES")

print(
    "H0: The mean glucose level is the same "
    "across all smoking-status groups."
)

print(
    "H1: At least one smoking-status group "
    "has a different mean glucose level."
)

# Significance level
alpha = 0.05

print(f"\nSignificance Level (α): {alpha}")

# 3. ONE-WAY ANOVA
f_statistic, p_value = stats.f_oneway(
    *smoking_groups
)

print("\nANOVA RESULTS")
print("-------------------------------------")

print(f"F-statistic: {f_statistic:.4f}")
print(f"p-value: {p_value:.6f}")


print("\nDECISION")

if p_value < alpha:

    print("Reject H0.")

else:

    print("Fail to reject H0.")


print("\nCONCLUSION")

if p_value < alpha:

    print(
        "There is statistically significant evidence "
        "that mean glucose level differs among "
        "at least some smoking-status groups."
    )

else:

    print(
        "There is not enough statistical evidence "
        "to conclude that mean glucose level differs "
        "among the smoking-status groups."
    )
```

```python
# PHASE 10 — CALCULUS

x = df_clean["age"]
y = df_clean["avg_glucose_level"]

# 1. Define the Function
slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)

def glucose_function(age):
    return slope * age + intercept

print("1. FUNCTION")
print("----------------------------------------")

print(
    f"f(x) = {slope:.4f}x + {intercept:.4f}"
)

age_example = 50

glucose_prediction = glucose_function(age_example)

print(
    f"\nWhen age = {age_example}:"
)

print(
    f"Predicted average glucose level = "
    f"{glucose_prediction:.2f}"
)



rate_of_change = slope

print("\n2. RATE OF CHANGE")
print("----------------------------------------")

print(
    f"Rate of Change = {rate_of_change:.4f}"
)

print(
    f"This means that for each 1-year increase in age, "
    f"the predicted average glucose level changes by "
    f"{rate_of_change:.4f} units according to the fitted model."
)


derivative = slope

print("\n3. DERIVATIVE")
print("----------------------------------------")

print(
    f"f'(x) = {derivative:.4f}"
)

print(
    "The derivative represents the rate of change "
    "of average glucose level with respect to age."
)


# 4. Explain How One Variable Changes When Another Changes
print("\n4. RELATIONSHIP BETWEEN THE VARIABLES")
print("----------------------------------------")

if derivative > 0:

    print(
        "The derivative is positive."
    )

    print(
        "Therefore, the model indicates that average glucose "
        "level tends to increase as age increases."
    )

elif derivative < 0:

    print(
        "The derivative is negative."
    )

    print(
        "Therefore, the model indicates that average glucose "
        "level tends to decrease as age increases."
    )

else:

    print(
        "The derivative is approximately zero."
    )

    print(
        "Therefore, the model indicates little linear change "
        "in average glucose level as age changes."
    )


print("\nFINAL INTERPRETATION")
print("----------------------------------------")

print(
    f"The fitted function is "
    f"f(x) = {slope:.4f}x + {intercept:.4f}."
)

print(
    f"The derivative is f'(x) = {derivative:.4f}, "
    f"which represents the estimated rate of change in "
    f"average glucose level for a one-year change in age."
)

print(
    "This describes a statistical relationship in the dataset "
    "and does not establish a causal relationship between age "
    "and glucose level."
)
```

```python
# PHASE 12 — FINAL FINDINGS & CONCLUSION

print("FINAL FINDINGS AND CONCLUSION")
print("=" * 70)


print("\n1. MAIN FINDINGS")

print(
    f"The cleaned dataset contains "
    f"{df_clean.shape[0]} records and "
    f"{df_clean.shape[1]} variables."
)

final_stroke_probability = (
    (df_clean["stroke"] == 1).sum()
    / len(df_clean)
)

print(
    f"The probability of stroke is "
    f"{final_stroke_probability:.2%}."
)

print(
    f"The mean age is "
    f"{df_clean['age'].mean():.2f} years."
)

print(
    f"The mean average glucose level is "
    f"{df_clean['avg_glucose_level'].mean():.2f}."
)

print(
    f"The mean BMI is "
    f"{df_clean['bmi'].mean():.2f}."
)


print("\n2. IMPORTANT STATISTICAL RESULTS")

print(
    f"Age: Mean = {df_clean['age'].mean():.2f}, "
    f"Median = {df_clean['age'].median():.2f}"
)

print(
    f"Average Glucose Level: "
    f"Mean = {df_clean['avg_glucose_level'].mean():.2f}, "
    f"Median = {df_clean['avg_glucose_level'].median():.2f}"
)

print(
    f"BMI: Mean = {df_clean['bmi'].mean():.2f}, "
    f"Median = {df_clean['bmi'].median():.2f}"
)

print("\n3. SIGNIFICANT RELATIONSHIPS")

final_hypertension_patients = df_clean[
    df_clean["hypertension"] == 1
]

final_p_stroke_given_hypertension = (
    (final_hypertension_patients["stroke"] == 1).sum()
    / len(final_hypertension_patients)
)

final_heart_disease_patients = df_clean[
    df_clean["heart_disease"] == 1
]

final_p_stroke_given_heart_disease = (
    (final_heart_disease_patients["stroke"] == 1).sum()
    / len(final_heart_disease_patients)
)

final_correlation = df_clean[
    ["age", "avg_glucose_level"]
].corr().iloc[0, 1]

print(
    f"P(Stroke | Hypertension) = "
    f"{final_p_stroke_given_hypertension:.2%}"
)

print(
    f"P(Stroke | Heart Disease) = "
    f"{final_p_stroke_given_heart_disease:.2%}"
)

print(
    f"Correlation between age and glucose = "
    f"{final_correlation:.3f}"
)

print("\n4. HYPOTHESIS TESTING RESULTS")

final_stroke_group = df_clean[
    df_clean["stroke"] == 1
]["avg_glucose_level"]

final_no_stroke_group = df_clean[
    df_clean["stroke"] == 0
]["avg_glucose_level"]

final_t_statistic, final_hypothesis_p_value = stats.ttest_ind(
    final_stroke_group,
    final_no_stroke_group,
    equal_var=False
)

print(
    f"t-statistic = "
    f"{final_t_statistic:.4f}"
)

print(
    f"p-value = "
    f"{final_hypothesis_p_value:.6f}"
)

if final_hypothesis_p_value < 0.05:

    print("Decision: Reject H0.")

    print(
        "Conclusion: There is statistically significant evidence "
        "that mean glucose level differs between stroke and "
        "no-stroke patients."
    )

else:

    print("Decision: Fail to reject H0.")

    print(
        "Conclusion: There is not enough evidence to conclude "
        "that mean glucose level differs between stroke and "
        "no-stroke patients."
    )


print("\n5. ANOVA RESULTS")

final_smoking_groups = [
    group["avg_glucose_level"].dropna()
    for name, group
    in df_clean.groupby("smoking_status")
]

final_f_statistic, final_anova_p_value = stats.f_oneway(
    *final_smoking_groups
)

print(
    f"F-statistic = "
    f"{final_f_statistic:.4f}"
)

print(
    f"p-value = "
    f"{final_anova_p_value:.6f}"
)

if final_anova_p_value < 0.05:

    print("Decision: Reject H0.")

    print(
        "Conclusion: At least one smoking-status group "
        "has a different mean glucose level."
    )

else:

    print("Decision: Fail to reject H0.")

    print(
        "Conclusion: There is not enough evidence that "
        "the mean glucose levels differ among smoking-status groups."
    )


print("\n6. CALCULUS FINDINGS")

final_x = df_clean["age"]
final_y = df_clean["avg_glucose_level"]

final_slope, final_intercept, _, _, _ = stats.linregress(
    final_x,
    final_y
)

print(
    f"Linear model: "
    f"Glucose = {final_slope:.4f} × Age + "
    f"{final_intercept:.4f}"
)

print(
    f"Derivative / Rate of Change = "
    f"{final_slope:.4f}"
)

if final_slope > 0:

    print(
        "Average glucose level tends to increase "
        "as age increases."
    )

elif final_slope < 0:

    print(
        "Average glucose level tends to decrease "
        "as age increases."
    )

else:

    print(
        "There is little linear change in glucose "
        "level as age changes."
    )


print("\n7. FINAL CONCLUSION")

print(
    "The project applied mathematical and statistical "
    "methods to analyze the healthcare stroke dataset."
)

print(
    "The analysis covered data preparation, descriptive "
    "statistics, distribution analysis, linear algebra, "
    "probability, sampling, confidence intervals, "
    "hypothesis testing, ANOVA, calculus, and visualization."
)

print(
    "The results provide insights into stroke probability, "
    "patient characteristics, relationships between variables, "
    "differences between groups, and the relationship between "
    "age and average glucose level."
)

print(
    "The findings describe relationships observed in the dataset "
    "and do not establish causal relationships."
)


print("\n8. FUTURE IMPROVEMENTS")

print(
    "Future analysis could use larger and more representative "
    "datasets and apply additional statistical and predictive "
    "methods to obtain stronger and more generalizable conclusions."
)
```
